```python
# =============================================================================
# CELULA 2 — CARREGAR AS BASES DE DADOS REAIS
# =============================================================================
# Carregando as bases diretamente da pasta local 'dados/' agora que ela foi enviada

from pathlib import Path

def achar_pasta_dados() -> Path:
    for candidato in ["dados", "../dados", "../../dados", "/content/dados"]:
        p = Path(candidato)
        if p.is_dir() and any(p.glob("*.csv")):
            return p
    return Path("dados")

def carregar(nome_arquivo: str) -> pd.DataFrame:
    caminho = achar_pasta_dados() / nome_arquivo
    if caminho.exists():
        df = pd.read_csv(caminho)
        print(f"[arquivo] {nome_arquivo} -> {df.shape[0]} linhas x {df.shape[1]} colunas")
        return df
    # ---- plano B: bases embutidas ----
    print(f"[sklearn] {nome_arquivo} nao encontrado em dados/ — usando a base embutida")
    from sklearn.datasets import load_breast_cancer, load_diabetes, load_digits, load_wine
    mapa = {"cancer_mama.csv": load_breast_cancer,
            "diabetes_progressao.csv": load_diabetes,
            "digitos.csv": load_digits,
            "vinhos.csv": load_wine}
    if nome_arquivo not in mapa:
        raise FileNotFoundError(
            f"A base '{nome_arquivo}' nao esta na pasta dados/. "
            "Baixe a pasta 'dados' que acompanha este material."
        )
    return mapa[nome_arquivo](as_frame=True).frame

# Carregando as cinco bases com prioridade para os arquivos locais
cancer   = carregar("cancer_mama.csv")
diabetes = carregar("diabetes_progressao.csv")
digitos  = carregar("digitos.csv")
vinhos   = carregar("vinhos.csv")
atacado  = carregar("clientes_atacado.csv")

print("\nResumo das bases carregadas:")
for nome, base in [("cancer", cancer), ("diabetes", diabetes), ("digitos", digitos),
                   ("vinhos", vinhos), ("atacado", atacado)]:
    print(f"  {nome:9s} -> {base.shape[0]:>5} linhas x {base.shape[1]:>3} colunas")

display(atacado.head())
```

O **K-Means** automatiza exatamente isso, em 4 passos que se repetem:

decisão de marketing.

Três ferramentas para responder "quantos grupos?" — e o que cada uma realmente mede:

* **Inércia (método do cotovelo / *elbow*)** = soma das distâncias de cada cliente ao centro do seu
  grupo. Quanto mais grupos, menor a inércia (no limite, cada cliente é seu próprio grupo e a inércia

  **deixa de cair rápido**. É subjetivo, mas é um bom primeiro olhar.
* **Silhueta (*silhouette*)** = de −1 a +1. Para cada cliente, compara: **(a)** a distância média para
  os do seu próprio grupo e **(b)** a distância para o grupo vizinho mais próximo. Se ele está muito
  mais perto dos seus, nota alta. **Acima de ~0,5** é estrutura razoável; **abaixo de 0,25** é
  praticamente nublado.
* **Davies-Bouldin** = o "inverso" da silhueta: **quanto menor, melhor**. Mede o quanto os grupos são
  compactos e separados entre si.
* **Calinski-Harabasz** = razão entre separação e coesão; **quanto maior, melhor**.

**A grande armadilha que esta questão revela:** a silhueta pode ser **alta** quando existe um grupo
minúsculo isolado (foi o caso do cluster de 2 na Questão 1). Número bom com grupo de 2 pessoas não
é segmentação — é estatística sendo enganada. **Sempre olhe o tamanho dos grupos junto com a métrica.**

* Ele procura **novos eixos** (chamados *componentes principais*) que são **combinações** das colunas
  originais. O primeiro eixo é a direção onde os dados **mais variam**; o segundo, a direção que
  **mais explica do que sobrou**; e assim por diante.
* **Variância explicada** = quanto da "informação" cada eixo carrega. Se 2 eixos explicam 22% da
  variância, então 78% se perde — mas, como veremos, **22% basta para enxergar a estrutura**.
* **Por que funciona?** Porque as colunas são **redundantes**: pixels vizinhos acendem juntos.
  Redundância = dá para resumir.

**Cuidado:** PCA é **linear**. Se a estrutura dos dados for uma curva (como um "S"), o PCA não
consegue "desdobrar" — para isso existem técnicas não lineares (t-SNE, UMAP).

<div class="card c-leigo"><h3>💡 Explicação para leigos — o conceito em português claro</h3>
O K-Means te dá **fotos** de uma divisão: "com 3 grupos, é assim". O hierárquico te dá o **filme
completo**, do começo ao fim:

* **Aglomerativo (bottom-up):** cada amostra começa sozinha. A cada passo, as **duas** coisas mais
  parecidas se juntam. Repete até sobrar um único grupo.
* **Dendrograma:** o desenho dessa árvore. A **altura** da linha horizontal onde dois ramos se unem
  mostra o quão diferentes eles são: **junção baixa = muito parecidos**; **junção alta = pouco
  parecidos**.
* **Onde cortar?** Imagine uma régua horizontal cruzando a árvore: quantas linhas verticais ela
  atravessa, quantos grupos você tem. O melhor corte costuma ser logo **abaixo do salto maior**
  (a maior "linha vertical sem junção").
* **Ward:** a regra de junção que minimiza o aumento da variância interna — tende a formar grupos
  do **mesmo tamanho** e é a melhor opção padrão.
* **Comparar com `complete` e `average`:** cada regra de junção ("linkage")produz uma árvore
  diferente. `single` gera o *efeito corrente* (grupos que viram uma fila enorme); `average` e
  `complete` ficam no meio-termo.

**Como medir se o agrupamento "acertou"?** Com o **ARI** (*Adjusted Rand Index*): 1,0 = concordância
perfeita com os rótulos verdadeiros; 0,0 = concordância que se obteria por acaso. Ele é ajustado
justamente para que o acaso dê zero.

em frescos). A diretoria quer uma **lista de auditagem**: quem investigar primeiro?

**Três hipóteses de negócio para um cliente atípico:**
1. **Grande conta** (cliente estratégico que merece atendimento dedicado);
2. **Erro de cadastro/digitação** (um zero a mais no pedido);
3. **Fraude** (compra atípica em nome de outro cliente).

**O problema.** Com 440 clientes e 6 categorias, inspecionar manualmente é inviável — e "suspeito"
é subjetivo. Você vai usar **Isolation Forest** (*Floresta de Isolamento*), um dos algoritmos de
detecção de anomalias mais usados em **mercados financeiros, cibersegurança e manutenção preditiva**.

**Por que não basta olhar a média?** Porque a anomalia pode estar na **combinação**: um cliente com
gasto normal em cada categoria isoladamente, mas com um *mix* impossível (ex.: 45 mil em delicatessen
e 239 em detergentes).
</div>

<div class="card c-leigo"><h3>💡 Explicação para leigos — o conceito em português claro</h3>
O **Isolation Forest** tem uma das ideias mais elegantes da computação:

> **Anomalias são fáceis de ISOLAR.**

Imagine um jogo de 20 perguntas para separar um ponto dos outros: você faz cortes aleatórios
("gasta mais de 5.000 em frescos?" / "gasta menos de 800 em mercearia?"). Um **ponto normal** está no
meio da nuvem e precisa de **muitos** cortes para ficar sozinho. Um **ponto esquisito** está no canto
e se separa em **poucos** cortes.

O algoritmo mede exatamente isso: **quantos cortes foram necessários para isolar cada ponto**.
Poucos cortes = suspeito.

* **`contamination`** = a "taxa de contaminação": **que % da base você espera que seja anômala?**
  É você quem define (aqui, 5%). É o hiperparâmetro mais importante e o mais político: define
  quantos casos o seu time terá de investigar.
* **`score_samples`** = o "grau de estranheza". **Quanto menor (mais negativo), mais anômalo.**
  Valores próximos de 0 são normais; em torno de −0,5 já é suspeito.
* **`predict`** devolve `1` (normal) ou `-1` (anomalia).

**A alternativa caseira:** o **z-score** — quantos desvios-padrão o valor está da média. A regra
`|z| > 3` é o clássico. Compare os dois métodos e responda: **eles pegam os mesmos clientes?**
(spoiler: não — e entender a diferença é o objetivo desta questão).
</div>

```python
#Resposta em código
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd

df_anom = atacado.copy()
gastos = ["frescos", "laticinios", "mercearia", "congelados", "detergentes_papel", "delicatessen"]
X_anom_scaled = StandardScaler().fit_transform(df_anom[gastos])

# (a) Floresta de Isolamento com taxa esperada de 5%
iso = IsolationForest(contamination=0.05, random_state=42, n_jobs=-1)
df_anom["anomalia"] = iso.fit_predict(X_anom_scaled)
df_anom["score"] = iso.score_samples(X_anom_scaled)

# (b) Listando os 5 principais suspeitos (menor score = mais isolado)
print("Top 5 Casos Mais Anômalo:")
top_anomalos = df_anom.nsmallest(5, "score")
print(top_anomalos[gastos + ["canal_nome", "score"]])

# (d) Comparação clássica Univariada com o Z-Score > 3
z_scores = np.abs((df_anom[gastos] - df_anom[gastos].mean()) / df_anom[gastos].std())
df_anom["outlier_z"] = (z_scores > 3).any(axis=1)

print("\nConfronto de Abordagens: Isolation Forest vs Z-Score Univariado:")
print(pd.crosstab(df_anom["anomalia"], df_anom["outlier_z"]))
```
