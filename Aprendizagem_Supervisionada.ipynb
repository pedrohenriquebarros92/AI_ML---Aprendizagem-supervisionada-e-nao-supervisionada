````python
# =============================================================================
# CELULA 2 — CARREGAR AS BASES DE DADOS REAIS
# =============================================================================
# As bases estao na pasta `dados/`, que acompanha este notebook.
# O codigo procura a pasta em alguns lugares possiveis (ao lado do notebook,
# uma ou duas pastas acima) e, se nao encontrar, cai no "plano B": busca a mesma
# base dentro do proprio scikit-learn (que tambem e real). Assim a aula never trava.

from pathlib import Path

def achar_pasta_dados() -> Path:
    """Procura a pasta de dados em caminhos comuns (Jupyter local, Colab, subpastas)."""
    for candidato in ["dados", "../dados", "../../dados"]:
        p = Path(candidato)
        if p.is_dir() and any(p.glob("*.csv")):
            return p
    return Path("dados")

def carregar(nome_arquivo: str) -> pd.DataFrame:
    """Le um CSV da pasta dados/. Se nao existir, usa a base equivalente do sklearn."""
    caminho = achar_pasta_dados() / nome_arquivo
    if caminho.exists():
        df = pd.read_csv(caminho)
        print(f"[arquivo] {nome_arquivo} -> {df.shape[0]} linhas x {df.shape[1]} colunas")
        return df
    # ---- plano B: bases identicas embutidas no scikit-learn ----
    print(f"[sklearn] {nome_arquivo} nao encontrado em dados/ — usando a base embutida")
    from sklearn.datasets import load_breast_cancer, load_diabetes, load_digits, load_wine
    mapa = {"cancer_mama.csv": load_breast_cancer,
            "diabetes_progressao.csv": load_diabetes,
            "digitos.csv": load_digits,
            "vinhos.csv": load_wine}
    if nome_arquivo not in mapa:
        raise FileNotFoundError(
            f"A base '{nome_arquivo}' nao esta na pasta dados/. "
            "Baixe a pasta 'dados' que acompanha este material."
        )
    return mapa[nome_arquivo](as_frame=True).frame


# Carregando as cinco bases usadas no material
cancer   = carregar("cancer_mama.csv")            # 569 exames (UCI Breast Cancer Wisconsin)
diabetes = carregar("diabetes_progressao.csv")    # 442 pacientes (Diabetes — Efron et al.)
digitos  = carregar("digitos.csv")                # 1.797 imagens 8x8 (OCR — Alpaydin & Kaynak)
vinhos   = carregar("vinhos.csv")                 # 178 amostras (UCI Wine)
atacado  = carregar("clientes_atacado.csv")       # 440 clientes (UCI Wholesale customers)

print("\nResumo das bases carregadas:")
for nome, base in [("cancer", cancer), ("diabetes", diabetes), ("digitos", digitos),
                   ("vinhos", vinhos), ("atacado", atacado)]:
    print(f"  {nome:9s} -> {base.shape[0]:>5} linhas x {base.shape[1]:>3} colunas")

display(cancer.head())      # display() mostra a tabela bonitinha no notebook
````

Conceitos desta questão, sem jargão:

* **Treino e teste** — você estuda por uma apostila (treino) e faz a prova com questões inéditas

* **Estratificar** — garantir que a prova tenha a mesma proporção de casos benignos e malignos

* **Precisão** — quando o modelo disse "maligno", quantas vezes ele estava certo.
  Precisão baixa gera pânico desnecessário e exames extras.
* **F1** — o equilíbrio entre precisão e recall, quando você não pode abrir mão de nenhum dos dois.

**O problema.** Dado os 64 pixels, dizer **qual dígito (0 a 9)** foi escrito. Agora não são 2 classes,
nem um número: são **10 classes**. É uma **classificação multiclasse**.

**Detalhe importante:** cada linha da tabela é uma imagem "desmontada" em 64 colunas
(`pixel_0_0` … `pixel_7_7`). Você vai precisar **remontar** a imagem para enxergá-la.

<div class="card c-leigo"><h3>💡 Explicação para leigos — o conceito em português claro</h3>
Pense em como você reconhece o "7" escrito por alguém: não é decorando o desenho exato, é comparando
com outros "7" que já viu. O **KNN (K-Nearest Neighbors / K vizinhos mais próximos)** faz exatamente
isso: guarda todos os exemplos e, quando chega um novo, pergunta aos *K* exemplos mais parecidos:
*"vocês são o quê?"* — e vai com a maioria.

* **K = 1** → pergunta a um só vizinho. Muito sensível a ruído (um desenho torto engana).
* **K = 3, 5, 7** → pergunta a vários. Mais estável, um pouco menos "detalhista".
* **Distância** = o quanto duas imagens são parecidas (aqui, a distância euclidiana entre os 64 pixels).
* **Matriz de confusão 10 × 10** mostra não só *quantos* erros houve, mas **quais dígitos se confundem**
  (por exemplo, 8 virando 1, ou 4 virando 7 — o que faz todo sentido visual).
* **Acurácia por classe (recall por dígito):** em OCR, um dígito errado em um valor de boleto custa
  caro — e o erro não é distribuído igualmente entre os dígitos.
* **Macro vs weighted average:** a média *macro* trata cada dígito igualmente; a *weighted* dá mais
  peso às classes com mais exemplos. Quando as classes têm tamanhos diferentes, elas divergem.
</div>

<div class="card c-leigo"><h3>💡 Explicação para leigos — o conceito em português claro</h3>
Até aqui nós "chutamos" os parâmetros do modelo. Nesta questão vamos **profissionalizar o processo**:

* **Pipeline** = uma linha de montagem lacrada. Você transforma e treina **dentro** dela, e por isso
  o escalonamento nunca "vaza" do teste para o treino (o erro mais comum e mais grave de ML).
* **Hiperparâmetro** = o botão que **você** ajusta, e não o modelo. Exemplos: a força da
  regularização (`C`), a profundidade da árvore, o número de vizinhos. É diferente de **parâmetro**,
  que é o que o modelo aprende sozinho (os pesos).
* **Validação cruzada (k-fold)** = em vez de uma única prova, fazer **5 provas**: a turma é dividida
  em 5 grupos; a cada rodada, 4 grupos estudam e 1 faz a prova — e depois você tira a média.
  Isso mostra se o resultado é **estável** ou se foi sorte de um sorteio feliz.
* **Estratificada** = cada uma das 5 provas mantém a proporção 67,7% / 32,3% das classes.
* **GridSearchCV** = testar **todas as combinações** de botões (ex.: `C` em [0,01; 0,1; 1; 10; 100])
  e escolher a que tem a melhor média na validação cruzada. É a "força bruta organizada".
* **Classes desbalanceadas** = quando 68% dos clientes são de um tipo. Um modelo que diz "é tudo
  Horeca" acerta 68% e **não serve para nada**. Por isso olhamos **F1** (e não só acurácia) e usamos
  estratificação em todas as divisões.
* **Vazamento de dados (data leakage)** = deixar a resposta escapar para o treino. Exemplo clássico:
  calcular a média de **toda** a base e só depois dividir treino/teste. Nunca faça isso.

Evita vazamento de dados e deixa o código limpo.</td></tr><tr><td>Padronização (StandardScaler)</td><td>Colocar todas as colunas na mesma régua: média 0 e desvio 1. Obrigatório para algoritmos baseados em distância.</td></tr><tr><td>Semente aleatória (random_state)</td><td>Número que fixa o sorteio. Com a mesma semente, todos na turma chegam ao mesmo resultado.</td></tr><tr><td>Vazamento de dados (data leakage)</td><td>Quando informação do teste escapa para o treino (ex.: escalar tudo antes de dividir). Mudar o limiar troca precisão por recall.</td></tr><tr><td>Cluster / grupo</td><td>Conjunto de registros parecidos entre si e diferentes dos outros.</td></tr><tr><td>Centróide</td><td>O 'registro médio' do cluster — a melhor foto-resumo do grupo.</td></tr><tr><td>Silhueta</td><td>Nota de −1 a +1 que diz se os grupos são coesos e separados. Acima de ~0,5 = estrutura razoável.</td></tr><tr><td>Inércia</td><td>Soma das distâncias de cada ponto ao centro do seu grupo. Usada no método do cotovelo.</td></tr><tr><td>ARI (Adjusted Rand Index)</td><td>Mede concordância entre duas partições. 1,0 = perfeita; 0,0 = igual ao acaso; pode ser negativo.</td></tr><tr><td>Outlier / anomalia</td><td>Ponto fora do padrão. Pode ser erro, fraude ou o seu melhor cliente — investigue antes de julgar.</td></tr><tr><td>LGPD</td><td>Lei Geral de Proteção de Dados (13.709/2018). Em ML, garante ao cidadão o direito de pedir revisão de decisões automatizadas (art. 20).</td></tr><tr><td>Viés (bias)</td><td>Tendência sistemática do modelo a errar para um lado — muitas vezes herdada de uma amostra que não representa a população.</td></tr><tr><td>Interpretabilidade</td><td>Capacidade de explicar **por que** o modelo decidiu algo. Essencial em crédito, saúde e decisões que afetam pessoas.</td></tr><tr><td>Classificação</td><td>Prever uma **categoria** (benigno/maligno, dígito 0–9, canal Horeca/Varejo).</td></tr><tr><td>Regressão</td><td>Prever um **número** (preço, índice clínico, tempo de entrega).</td></tr><tr><td>Regressão Logística</td><td>Apesar do nome, é **classificação**. Estima a probabilidade de cada classe e é ótima como primeiro modelo.</td></tr><tr><td>Árvore de decisão</td><td>Modelo em forma de perguntas encadeadas ('gasto > 5.000?'). Fácil de explicar, mas decora fácil.</td></tr><tr><td>Random Forest (Floresta Aleatória)</td><td>Várias árvores que votam. Mais estável que uma árvore só, e bem difícil de explicar.</td></tr><tr><td>KNN (K vizinhos mais próximos)</td><td>Classifica perguntando aos k exemplos mais parecidos. Exige escala e guarda todos os dados na memória.</td></tr><tr><td>SVM (Máquina de Vetores de Suporte)</td><td>Procura a fronteira mais larga possível entre as classes. Forte em bases pequenas e de muitas colunas.</td></tr><tr><td>MAE (erro médio absoluto)</td><td>Em média, de quanto o modelo errou — na unidade do problema (R$, pontos).</td></tr><tr><td>RMSE (raiz do erro quadrático médio)</td><td>Parecido com o MAE, mas pune muito os erros grandes.</td></tr><tr><td>R² (coeficiente de determinação)</td><td>Quanta variação do alvo o modelo explica. 1,0 = perfeito; 0 = igual à média; negativo = pior que a média.</td></tr><tr><td>Resíduo</td><td>Real − previsto. Analisar resíduos é o raio-X de um modelo de regressão.</td></tr><tr><td>Multicolinearidade</td><td>Quando duas colunas medem quase a mesma coisa; os coeficientes ficam instáveis e trocam de sinal.</td></tr><tr><td>Regularização (Ridge/Lasso)</td><td>Penalizar modelos 'exagerados' para evitar decorar. O `C` da regressão logística é o inverso da força da regularização.</td></tr><tr><td>GridSearchCV</td><td>Força bruta organizada: testa todas as combinações de hiperparâmetros e escolhe pela validação cruzada.</td></tr><tr><td>class_weight</td><td>Dá mais peso à classe minoritária. Sobe o recall, derruba a precisão — é uma decisão de negócio.</td></tr><tr><td>Curva ROC e AUC</td><td>ROC mostra o desempenho em todos os limiares; AUC resume: 0,5 = aleatório, 1,0 = perfeito.</td></tr><tr><td>Importância por permutação</td><td>Embaralhar uma coluna e ver quanto a métrica cai. Mede o quanto o modelo depende dela.</td></tr><tr><td>joblib</td><td>Biblioteca para gravar e carregar o modelo treinado em disco (sem re-treinar).</td></tr><tr><td>Probabilidade (predict_proba)</td><td>O grau de confiança do modelo. Permite criar filas de revisão humana nos casos duvidosos.</td></tr><tr><td>Dados desbalanceados</td><td>Quando uma classe tem muito mais exemplos que a outra. A acurácia vira uma armadilha.</td></tr></tbody></table>
